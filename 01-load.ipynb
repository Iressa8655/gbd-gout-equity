```python
import pandas as pd

# 載入 GBD 2023 痛風 DALYs
gbd = pd.read_csv(
    "data/raw/IHME-GBD_2023_DATA-54ccbeed-1/IHME-GBD_2023_DATA-54ccbeed-1.csv"
)

print("GBD shape:", gbd.shape)
print(gbd.dtypes)
gbd.head()
```

```python
gdp_raw = pd.read_csv(
    "data/raw/API_NY.GDP.PCAP.PP.CD_DS2_en_csv_v2_121708/API_NY.GDP.PCAP.PP.CD_DS2_en_csv_v2_121708.csv",
    skiprows=4
)

print("GDP shape:", gdp_raw.shape)
print(gdp_raw.columns.tolist()[:10])
gdp_raw.head(3)
```

```python
# Wide → Long format
gdp_long = gdp_raw.melt(
    id_vars=["Country Name", "Country Code"],
    value_vars=[str(y) for y in range(1990, 2024)],
    var_name="year",
    value_name="gdp_per_capita_ppp"
)

# year 欄位從文字轉成數字
gdp_long["year"] = gdp_long["year"].astype(int)

# 刪掉沒有 GDP 資料的行
gdp_long = gdp_long.dropna(subset=["gdp_per_capita_ppp"])

print("GDP long shape:", gdp_long.shape)
gdp_long.head()
```
